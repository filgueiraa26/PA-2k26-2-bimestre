```python

#Exercício 1 — Contagem

numero = int(input("Digite um número: "))

contador = 1

while contador <= numero:
    print(contador)
    contador = contador + 1
```

```python
#Exercício 2 — Contagem regressiva

numero = int(input("Digite um número: "))

while numero >= 0:
    print(numero)
    numero -= 1
```

```python
#Exercício 3 — Números pares

numero = 0

while numero <= 20:
    print(numero)
    numero = numero + 2
```

```python
#Exercicio 4 - Soma dos números

numero = int(input("Digite um número: "))

contador = 1
soma = 0

while contador <= numero:
    soma = soma + contador
    contador = contador + 1

print("Soma:", soma)
```

```python
#Exercício 5 — Tabuada

numero = int(input("Digite um número: "))

contador = 1

while contador <= 10:
    resultado = numero * contador
    print(numero, "x", contador, "=", resultado)
    contador = contador + 1
```

```python
#Exercício 6 — Senha

senha = input("Digite a senha: ")

while senha != "6742":
    print("Senha incorreta!")
    senha = input("Digite a senha: ")

print("Acesso concedido. Seja bem vindo!")
```

```python
#Exercício 7 — Somando até digitar zero

soma = 0

numero = int(input("Digite um número: "))

while numero != 0:
    soma = soma + numero
    numero = int(input("Digite um número: "))

print("Soma:", soma)
```
