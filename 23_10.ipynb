### MATCH:

```python
#1:
cor = 1
match cor:
    case 1:
        print("Cor azul")
    case _:
        print("Outra cor")
```

```python
#2:
numero = 2

match numero:
  case 1:
    print("Numero dm")
  case 2:
    print("Numero dois")
  case 3:
    print("Numero três")
```

```python
#3:
opcao = int(input("Digite 1, 2 ou 3: "))

match opcao:
    case 1:
        print("Água")
    case 2:
        print("Suco")
    case 3:
        print("Refrigerante")
    case _:
        print("Opção inválida")
```

```python
#4:
dia = input("Digite um dia da semana: ")

match dia:
    case "sábado" | "domingo":
        print("Final de semana!")
    case _:
        print("Dia de semana.")
```

```python
#5:
nota = input("Digite uma nota (A, B ou C): ")

match nota:
    case "A":
        print("Excelente")
    case "B":
        print("Bom")
    case "C":
        print("Precisa melhorar")
    case _:
        print("Nota inválida")
```

## RANGE:


```python
#1:
for numero in range(10):
    print(numero)
```

```python
#2:
for numero in range(1, 11):
    print(numero)
```

```python
#3:
for numero in range(5, 16):
    print(numero)
```

```python
#4:
for numero in range(0, 21, 2):
    print(numero)
```

```python
#5:
for numero in range(1, 16, 2):
    print(numero)
```

```python
#6:
for numero in range(10, 0, -1):
    print(numero)
```

```python
#7:
for numero in range(1, 11):
    print("3 x", numero, "=", 3 * numero)
```
