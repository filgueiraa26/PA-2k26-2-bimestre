```python
# Calculadora com as 4 operações
# Opção 1- Somar
# Opção 2- Subtrair
# Opção 3- Multiplicar
# Opção 4- Divisão
# Opção 5- Sair

opcao = 0
a = 0
b = 0

print ("============================")
print ("Opção 1- Soma")
print ("Opção 2- Subtrair")
print ("Opção 3- Multiplicar")
print ("Opção 4- Dividir")
print ("Opção 5- Sair")
print ("===========================")

while opcao != 5:
  print ("Opção 1- Soma")
  print ("Opção 2- Subtrair")
  print ("Opção 3- Multiplicar")
  print ("Opção 4- Dividir")
  print ("Opção 5- Sair")
  print ("==========================")

  opcao = int (input("digite uma opção de 1 a 5 para Sair: "))

  if opcao == 1:

    a = int (input("Digite um número: "))
    b = int (input("Digite outro número: "))

    print(f"total de soma igual a: {a + b}")

  elif opcao == 2:

    a = int (input("Digite um número: "))
    b = int (input("Digite outro número: "))

    print(f"total de subtração igual a: {a - b}")

  elif opcao == 3:

    a = int (input("Digite um número: "))
    b = int (input("Digite outro número: "))

    print(f"total de multiplicação igual a: {a * b}")

  elif opcao == 4:

    a = int (input("Digite um número: "))
    b = int (input("Digite outro número: "))

    print(f"total de divisão igual a: {a / b}")

  else:
    print("Até mais!!")
```
