```python
# Escrever um jogo de adivinhação. O usúario terá 3 chances para adivinhar um número de 0 a 10.
# O programa da uma mensagem se o palpite foi maior ou menos.
# O programa avisa quente ou frio.

import random

numero = random.randint(0, 10)

tentativas = 3

print("---------JOGO DE ADIVINHAÇÃO---------")
print("Tente adivinhar um número de 0 a 10!!")
print("-------------------------------------")
for i in range(tentativas):

    palpite = int(input("Digite seu palpite: "))


    if palpite == numero:
        print("Parabéns! Você acertou!")
        break

    else:
        if palpite > numero:
            print("Seu palpite foi MAIOR que o número.")

        else:
            print("Seu palpite foi MENOR que o número.")

        diferenca = abs(palpite - numero)

        if diferenca <= 2:
            print("Está QUENTE!")
        else:
            print("Está FRIO!")

        print("Tentativas restantes:", tentativas - i - 1)

if palpite != numero:
    print("Você perdeu!")
    print("O número era:", numero)
```
