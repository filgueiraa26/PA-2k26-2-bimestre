```python
#1- Dados Pessoais

#Entrada

Nome= str(input("Digite seu nome: "))
Idade= int(input("Digite sua idade: "))
Altura= float(input("Digite sua altura: "))

print(f"Ooi {Nome} ")
print(f"Você tem {Idade} anos.")
print(f"E tem {Altura} metros!")
```

```python
#2- Antecessor e Sucessor

numero = int(input("Digite um número inteiro: "))

# Exibe o número, seu antecessor e seu sucessor
print(f"Número informado: {numero}")
print(f"Antecessor: {numero - 1}")
print(f"Sucessor: {numero + 1}")
```

```python
#3- Operações Matemáticas

# Solicita dois números ao usuário
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))

# Calcula as operações
soma = num1 + num2
subtracao = num1 - num2
multiplicacao = num1 * num2

# Verifica se o segundo número é diferente de zero
if num2 != 0:
    divisao = num1 / num2
else:
    divisao = "Não é possível dividir por zero."

# Exibe os resultados
print("\n--- Resultados ---")
print(f"Soma: {soma}")
print(f"Subtração: {subtracao}")
print(f"Multiplicação: {multiplicacao}")
print(f"Divisão: {divisao}")
```

```python
#4- Média de Notas

# Solicita as três notas
nota1 = float(input("Digite a 1ª nota: "))
nota2 = float(input("Digite a 2ª nota: "))
nota3 = float(input("Digite a 3ª nota: "))

# Calcula a média
media = (nota1 + nota2 + nota3) / 3

# Verifica a situação do aluno
if media >= 7:
    situacao = "Aprovado"
elif media >= 5:
    situacao = "Recuperação"
else:
    situacao = "Reprovado"

# Exibe os resultados
print("\n--- Resultado ---")
print(f"Média: {media:.2f}")
print(f"Situação: {situacao}")
```

```python
#5- Conversão de Temperatura

# Solicita a temperatura em Celsius
celsius = float(input("Digite a temperatura em graus Celsius: "))

# Converte para Fahrenheit
fahrenheit = (celsius * 9 / 5) + 32

# Exibe o resultado
print(f"\nTemperatura em Celsius: {celsius:.1f}°C")
print(f"Temperatura em Fahrenheit: {fahrenheit:.1f}°F")
```

```python
#6- Área de um Retângulo

# Solicita a base e a altura do retângulo
base = float(input("Digite a base do retângulo: "))
altura = float(input("Digite a altura do retângulo: "))

# Calcula a área
area = base * altura

# Exibe o resultado
print(f"\nA área do retângulo é: {area}")
```

```python
#7- Número Par ou Ímpar

#Entrada

numero = int (input("Digite um número: "))

# Processamento / Saída

if numero % 2 == 0:

  print (f" o numero {numero} é par. ")

else:
  print (f"O numero {numero} é ímpar. ")
```

```python
#8- Maior Número


#Entrada

print("Digite três números para encontrar o maior:")
numero1 = int(input("Digite o primeiro número: "))
numero2 = int(input("Digite o segundo número: "))
numero3 = int(input("Digite o terceiro número: "))

maior_numero = numero1

if numero2 > maior_numero:
    maior_numero = numero2
if numero3 > maior_numero:
    maior_numero = numero3

print(f"O maior número digitado é: {maior_numero}")
```

```python
#9- Tabuada

#Entrada

numero = int(input("Digite um número para ver a tabuada: "))

print(f"Tabuada do {numero}:")
for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")
```

```python
#10- Calculadora Simples

#Entrada

print ("|------------CALCULADORA------------|")
print ("|                                   |")
print ("|---------------MENU----------------|")
print ("|Opção = 1 = Soma                   |")
print ("|Opção = 2 = Subtração              |")
print ("|Opção = 3 = Multiplicação          |")
print ("|Opção = 4 = Divisão                |")
print ("|-----------------------------------|")

# Entrada
numero = float (input ("Digite um número: "))
opcao  = int (input("Escolha a opção desejada: "))
numero1 = float (input("Digite um numero: "))

# Processamento / saída

if opcao == 1:
  print (f" A soma de {numero} e {numero1} é igual a {numero + numero1} :")

elif opcao == 2:
  print (f" A subtração de {numero} e {numero1} é igual a {numero - numero1} :")

else:
  print ("Opção Inválida")
```
