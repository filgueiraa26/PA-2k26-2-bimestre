```python
#1: Elabore um programa que exiba na tela os números de 1 até 20, #um em cada linha.


a = int(input("Digite um numero: "))
for i in range(0,a + 1):
  print(i)
```

```python
#2: Solicite ao usuário um número inteiro positivo.
#Utilizando um laço de repetição, calcule e exiba a soma de todos os números de 1 até o número informado.

a = int(input("Digite um número: "))
for i in range(1, a):
  a += i
print(a)
```

```python
#3: Solicite ao usuário um número inteiro e exiba sua tabuada de 1 até 10.

numero = int(input("Digite um número para ver a tabuada: "))

print(f"Tabuada do {numero}:")
for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")
```

```python
#4: Solicite ao usuário um número inteiro positivo e calcule seu fatorial utilizando um laço de repetição.

numero = int(input("Digite um número inteiro positivo: "))
fatorial = 1
for i in range(1, numero + 1):
    fatorial *= i
print(f"O fatorial de {numero} é: {fatorial}")
```

```python
#5: Solicite ao usuário o valor do saque e informe a menor quantidade possível de notas para realizar o pagamento.

valor = int(input("Digite o valor do saque: "))

nota = 100

while valor > 0:

    qtd = valor // nota

    if qtd > 0:
        print(f"{qtd} nota(s) de R${nota}")
        valor = valor % nota

    if nota == 100:
        nota = 50
    elif nota == 50:
        nota = 20
    elif nota == 20:
        nota = 10
    elif nota == 10:
        nota = 5
    elif nota == 5:
        nota = 2
    elif nota == 2:
        nota = 1

```
