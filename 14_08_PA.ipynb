```python
#3 - Maior de dois números

a = int(input("digite um número: "))
b = int(input("digite outro número: "))

if a > b:
  print(f"{a} é maior que {b}")
elif a == b:
  print(f"{a} é igual a {b}")
else:
  print(f"{b} é maior que {a}")
```

```python
#5 - Desconto na compra

valor = float(input("qual o valor de sua compra? "))
if valor >= 100:
  desc = valor * 0.1
  prec = valor - desc
  print(f"sua compra ganhou 10% de desconto e ficou {prec}")
elif valor >= 50 and valor < 100:
  desc = valor * 0.05
  prec = valor - desc
  print(f"sua compra ganhou 5% de desconto e ficou {prec}")
else:
  print(f"sua compra não ganhou desconto, e ficou {valor}")
```

```python
#6 - Positivo, negativo ou zero

nu = float(input("Digite um número: "))
if nu > 0:
  print (f"o número {nu} é positivo.")
elif nu == 0:
    print(f"o numero {nu} é igual a zero")
else:
    print(f"o numero {nu} é negativo")
```

```python
#7 - Classificação de idade

idade = int(input("digite sua idade: "))
if idade <= 12:
  print("você é uma criança")
elif idade >= 13 and idade <= 17:
  print("você é um adolescente")
elif idade >= 18 and idade <= 59:
  print("você é um adulto")
else:
  print("você é um idoso")
```

```python
#8 - Média de duas notas

nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))
# Calcula a média
media = (nota1 + nota2 + nota3) / 3

print(f"Média: {media:5}")

if media >= 7:
    print("Situação: Aprovado")
elif media >= 5:
    print("Situação: Recuperação")
else:
    print("Situação: Reprovado")
```

```python
#10 - Calculadora de IMC

# Entrada de dados
peso = float(input("Digite o peso em kg: "))
altura = float(input("Digite a altura em metros: "))

# Cálculo do IMC
imc = peso / (altura * altura)

# Exibição do resultado do IMC formatado com duas casas decimais
print(f"\nSeu IMC é: {imc:.2f}")

# Classificação do IMC com base nas faixas informadas
if imc < 18.5:
    print("Classificação: Abaixo do peso")
elif 18.5 <= imc <= 24.9:
    print("Classificação: Peso normal")
elif 25.0 <= imc <= 29.9:
    print("Classificação: Sobrepeso")
else:
    print("Classificação: Obesidade")
```
